```python
import asyncio

from IPython.display import display

from test_geo_host import chart_fixture

assert asyncio.get_running_loop().is_running()
source, chart = chart_fixture()
widget = chart.widget()
display(widget)
```

```python
old = widget._adapter._frame
assert old is not None
assert old.data.record(0)["feature_id"] == (1 << 64) - 1
assert old.data.identity["time"]["instant"] == -(1 << 63)
widget.close()
for _ in range(100):
    if not widget._adapter.mounted:
        break
    await asyncio.sleep(0.02)
assert not widget._adapter.mounted
try:
    _ = old.data
    raise AssertionError("native frame retained after browser ACK")
except RuntimeError:
    pass
source.close()
print("XYG geographic browser release acknowledged")
```
