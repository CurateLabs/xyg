```python
import asyncio

from IPython.display import display

from geo_live_hierarchy_fixture import live_hierarchy_chart_fixture

assert asyncio.get_running_loop().is_running()
source, scope, lane, frame, chart = live_hierarchy_chart_fixture()
widget = chart.widget(frame=frame, selected_scope=scope, hierarchy_lane=lane)
frame.close()
display(widget)
```

```python
assert widget._adapter._sequence == 4
await asyncio.wrap_future(
    widget.update(
        operation=8,
        args=[0, 0],
        sequence=5,
        camera_revision=5,
        time_revision=5,
        state_revision=1,
        time={"kind": 0},
    )
)
assert widget._adapter._frame.data.identity["time"]["kind"] == 0
assert widget._adapter._sequence == 5
widget.close()
for _ in range(200):
    if not widget._adapter.mounted:
        break
    await asyncio.sleep(0.05)
assert not widget._adapter.mounted
lane.close()
scope.close()
source.close()
print("XYG selected hierarchy browser release acknowledged")
```
