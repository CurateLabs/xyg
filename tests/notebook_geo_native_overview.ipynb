```python
import asyncio

from IPython.display import display

from test_geo_native_overview_host import fixture

assert asyncio.get_running_loop().is_running()
source, seed, index, chart = fixture()
widget = chart.widget()
display(widget)
```

```python
assert widget._adapter._sequence == 3
await asyncio.wrap_future(
    widget.update(
        operation=8,
        args=[0, 0],
        sequence=4,
        camera_revision=4,
        time_revision=4,
        state_revision=1,
        time={"kind": 0},
    )
)
assert widget._adapter._frame.data.final is False
assert widget._adapter._sequence == 4 and index.current is None
source.close()
seed.close()
index.close()
widget.close()
for _ in range(200):
    if not widget._adapter.mounted:
        break
    await asyncio.sleep(0.05)
assert not widget._adapter.mounted
print("XYG overview browser release acknowledged")
```
